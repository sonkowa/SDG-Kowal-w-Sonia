```python
import geopandas as gpd

path_files = "C:\\Users\\Sonia\\Documents\\studia\\semestr 5\\Struktury danych geoprzestrzennych\\SDG-Kowal-w-Sonia\\scripts\\temp\\2310_dane\\"

pasy = gpd.read_file(path_files+"pl_pasy_ukształtowania.geojson")

cities = gpd.read_file(path_files+"pl_miasta_wojewódzkie.geojson")

rivers = gpd.read_file(path_files+"pl_rzeki.geojson")
```

```python
#Dla warstwy z pasami ukształtowania policz ich powierzchnie w m2, korzystając z układu epsg:2180

print(pasy.crs)

pasy_2180 = pasy.to_crs("EPSG:2180")

print(f"Pasy mają powierzchnie {round(pasy_2180.iloc[0].geometry.area,2)} metrów ^2")


```

```python
#Stwórz obiekt Bounding Box dla warstwy z miastami wojewódzkimi, a następnie oblicz jego obwód

cities_2180 = cities.to_crs("EPSG:2180")

minx, miny, maxx, maxy = cities_2180.total_bounds
obwod = round(2*((maxx - minx) + (maxy - miny)),2)

print(f"Obwód Bbox dla warstwy z miastami wynosi: {obwod}")
```

```python
#Dla warstwy z rzekami stwórz nową kolumnę w której wyliczysz liczbę linii (LineString) znajdujących się wewnątrz. 
#Analogicznie stwórz następnie drugą kolumnę w której wyliczysz łączną liczbę punktów składających się na geometrię
import shapely as shp

rivers_2180 = rivers.to_crs("EPSG:2180")

rivers_2180["długość_linii"] = (rivers_2180.geometry.length / 1000).round(2)

rivers["liczba_pkt"] = rivers_2180['geometry'].apply(shp.count_coordinates)

rivers_2180
```
