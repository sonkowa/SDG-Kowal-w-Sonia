```python
import numpy as np
```

```python
#Stwórz losową 1-wymiarową numpy array o długości 15 elementów. Następnie oblicz jej minimum, maksimum oraz odchylenie standardowe.
array_random = np.random.randint(1, 100, 15)
print(array_random)


print(f"Minimum = {np.min(array_random)}\nMaximum = {np.max(array_random)}\nOdchylenie = {np.std(array_random)}")
```

```python
#Stwórz losową tablicę 2-wymiarową numpy array 4 wiersze x 4 kolumny, 
#a następnie oblicz sumę wartości dla każdej kolumny oraz dla każdego wiersza.

random_arr2 = np.random.randint(100, size = (4,4))
print(random_arr2)

print(f"Suma wierszy = {random_arr2.sum(axis=1)}")
print(f"Suma kolumn = {random_arr2.sum(axis=0)}")
```

```python
# Stwórz tablicę 2-wymiarową 5 wierszy x 5 wierszy, w której każda wartość będzie o 3 większa od poprzedniej. 
# Podaj indeksy każdej wartości parzystej większej od 10.

array_arange = np.arange(0,75,3).reshape(5,5)

print(array_arange)

print(np.where((array_arange < 10) % 2 == 0))
```
